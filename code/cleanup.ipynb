## Limpieza de datos

### Detección y tratamiento de duplicados

Primero, revisamos si hay filas duplicadas exactas. Luego, hay que analizar que un mismo video_id puede aparecer en varias fechas de tendencia (eso no es duplicado, es una serie temporal). Pero si hay dos filas con exactamente el mismo video_id y la misma trending_date, sí sería duplicado. Así que buscamos eso también.

```python
# Duplicados exactos (todas las columnas)
dups_exact = df.duplicated().sum()
print(f"Filas exactamente duplicadas: {dups_exact}")

# Duplicados por video_id + trending_date (combinación única lógica)
dups_logic = df.duplicated(subset=['video_id', 'trending_date']).sum()
print(f"Filas con mismo video_id y trending_date: {dups_logic}")
```

Como no existen filas duplicadas exactas ni filas con el mismo video_id y trending_date (la única otra forma que filas pueden ser duplicadas), entonces continuamos con la limpieza de nulos

### Detección de nulos estándar

```python
# Conteo de nulos estándar
null_counts = df.isnull().sum()
null_counts = null_counts[null_counts > 0].sort_values(ascending=False)
print("Valores nulos estándar (NaN) por columna:")
print(null_counts)
```

### Detección de "nulos encubiertos" en columnas de tipo string

A veces, muchos datasets contienen nulos encubiertos que una busqueda de nulos estándar no puede captar,

Entonces, necesitamos definir una lista negra de patrones que consideramos representan "vacío" o "sin información".

En el análisis inicial, se encontró en el dataset que en la columna `tags` hay un `"[none]"` que representa nulidad.

De todas maneras, buscaremos otras formas de nulidad comunes que pueden ser parte del dataset.

```python
# Lista de patrones que representan ausencia de información
null_patterns = [
    '[none]', 'none', 'null', 'nan', 'n/a', 'na', 
    '""', "''"  # cadenas vacías literales (aunque pandas las lee como '' )
]

def detect_hidden_nulls(df, patterns):
    """
    Recorre las columnas de tipo string y cuenta 
    cuántos valores coinciden con los patrones (case-insensitive).
    """
    hidden_counts = {}
    for col in df.select_dtypes(include=['str']).columns:
        # Convertir a string, quitar espacios, pasar a minúsculas
        col_clean = df[col].str.strip().str.lower()
        # Contar cuántos valores están en la lista de patrones
        mask = col_clean.isin(patterns)
        count = mask.sum()
        if count > 0:
            # Además, mostrar ejemplos de los valores encontrados
            examples = df.loc[mask, col].head(3).tolist()
            hidden_counts[col] = {
                'count': count,
                'examples': examples
            }
    return hidden_counts

# Ejecutar la detección
hidden_nulls = detect_hidden_nulls(df, null_patterns)
print("\n--- Nulos encubiertos detectados ---")
for col, info in hidden_nulls.items():
    print(f"Columna '{col}': {info['count']} valores sospechosos. Ejemplos: {info['examples']}")
```

Parece que los sospechosos usuales no se encuentran en el dataset, y solo hay el `"[none]"` que ya habíamos detectado, asi que lo que vamos a hacer es unificar los nulls para que asi se puedan imputar más fácilmente 

```python
def unify_hidden_nulls(df, patterns):
    """
    Reemplaza los patrones de nulidad en columnas string por np.nan.
    """
    df_clean = df.copy()
    for col in df_clean.select_dtypes(include=['str']).columns:
        # Aplicar strip y lower para comparar, pero mantener el original para reemplazo
        mask = df_clean[col].str.strip().str.lower().isin(patterns)
        df_clean.loc[mask, col] = np.nan
    return df_clean

# Aplicar la unificación
df = unify_hidden_nulls(df, null_patterns)

# Verificar nuevamente los nulos estándar (ahora deberían haber aumentado)
new_null_counts = df.isnull().sum()
new_null_counts = new_null_counts[new_null_counts > 0].sort_values(ascending=False)
print("\n--- Nuevos conteos de nulos (después de unificar) ---")
print(new_null_counts)
```

### Limpieza de espacios en blanco

A veces las cadenas tienen espacios al inicio o final que dificultan el análisis (ej. " Music "). Es buena práctica limpiarlos:

```python
# Limpiar espacios en blanco al inicio/final de todas las columnas de texto
for col in df.select_dtypes(include=['str']).columns:
    if df[col].dtype == 'str':
        df[col] = df[col].str.strip()
```

### Reporte de calidad final

```python
print("\n========== INFORME DE CALIDAD FINAL ==========")
print(f"Total de filas: {df.shape[0]}")
print(f"Total de columnas: {df.shape[1]}")

print(f"\nFilas exactamente duplicadas: 0")
print(f"Filas con mismo video_id y trending_date: 0")

# Porcentaje de nulos por columna (solo las que tienen)
null_percent = (df.isnull().sum() / len(df)) * 100
null_percent = null_percent[null_percent > 0].sort_values(ascending=False)
print("\nPorcentaje de nulos por columna:")
print(null_percent.round(2).astype(str) + '%')
```

### Imputación de datos nulos

Como los datos que son nulos son de tipo string, son porcentajes bajos y vamos a usar estas columnas principalmente para ingeniería de características (extraer longitud, contar etiquetas, o tal vez usar texto), se imputará rellenando con una cadena vacía ('')

#### ¿Por qué?

* Se preservan el 100% de los datos para las demás variables numéricas (vistas, likes, fecha, etc.) que son las más importantes para el modelo predictivo.

* Es semánticamente correcto:

   * Un tags vacío significa literalmente que el video no tiene etiquetas, lo cual es un dato válido. Si calculamos el número de tags (len(tags.split())), que sea 0 es correcto.

   * Una description vacía significa que el creador no escribió descripción. Si calculamos la longitud (len(description)), que sea 0 también es correcto.

* Se evita perder ~3,500 filas innecesariamente, lo cual es valioso para el entrenamiento del modelo (más datos = mejor generalización).

```python
# Imputar los nulos en tags y description con string vacío
df['tags'] = df['tags'].fillna('')
df['description'] = df['description'].fillna('')

# Verificar que ya no queden nulos
print(df[['tags', 'description']].isnull().sum())
# Debería dar 0 para ambas.
```

## Detección de outliers

Aunque la detección de outliers es parte de la limpieza de datos, le daremos un espacio aparte, porque se presentan resultados más interesantes y requieren de análisis muy cuidadoso para determinar si se deben imputar

### Visualización para detectar outliers

```python
# Seleccionar columnas numéricas (excluyendo lat/lon porque son coordenadas fijas)
num_cols = ['views', 'likes', 'dislikes', 'comment_count']

# Configurar el tamaño de la figura
fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(15, 10))
axes = axes.flatten()

for i, col in enumerate(num_cols):
    # Boxplot
    sns.boxplot(data=df, y=col, ax=axes[i], color='skyblue')
    axes[i].set_title(f'Boxplot de {col}')
    axes[i].set_ylabel(col)
    
# Ocultar el último subplot si sobra
if len(num_cols) < 6:
    axes[-1].set_visible(False)

plt.tight_layout()
plt.show()
```

### Detección sistemática de outliers mediante IQR (Rango Intercuartílico)

```python
def detect_outliers_iqr(df, column):
    """
    Detecta outliers en una columna numérica usando el método IQR.
    Retorna la máscara booleana y los límites inferior/superior.
    """
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Máscara de outliers (True = es outlier)
    mask = (df[column] < lower_bound) | (df[column] > upper_bound)
    
    return mask, lower_bound, upper_bound

# Aplicar a las columnas numéricas (excluyendo Likes/dislikes por ahora, lo tratamos aparte)
cols_to_check = ['views', 'likes', 'dislikes', 'comment_count']

outliers_summary = {}
for col in cols_to_check:
    mask, lb, ub = detect_outliers_iqr(df, col)
    outliers_summary[col] = {
        'count': mask.sum(),
        'percentage': (mask.sum() / len(df)) * 100,
        'lower_bound': lb,
        'upper_bound': ub
    }
    print(f"Columna '{col}': {mask.sum()} outliers ({outliers_summary[col]['percentage']:.2f}%)")
    print(f"  Límite inferior: {lb:.2f}, Límite superior: {ub:.2f}")
```

Para la detección de valores atípicos se aplicó el método del Rango Intercuartílico (IQR). Se identificaron valores extremos en el límite superior de las columnas views, likes, dislikes y comment_count, lo cual es esperado dado que la distribución de estas variables sigue una ley de potencia (cola larga), característica intrínseca de los fenómenos virales en redes sociales. Por este motivo, se decidió NO aplicar ninguna técnica de capping o eliminación sobre estos valores extremos, ya que representan la señal de interés para el cliente (viralidad). Solo se verificó que no existieran valores negativos (imputándolos a 0 en caso de existir, aunque no fue necesario). Para el modelado predictivo, la variable views será transformada mediante logaritmo natural (log1p) para estabilizar la varianza, preservando íntegramente el orden y la magnitud de los valores extremos.

```python
# SOLO por seguridad, si hubiera algún valor negativo (no debería), lo llevamos a 0.
# Pero NO tocamos los límites superiores.
cols_to_keep_original = ['views', 'likes', 'dislikes', 'comment_count']

for col in cols_to_keep_original:
    # Si por casualidad hay algún negativo (raro), lo fijamos a 0
    df[col] = df[col].clip(lower=0)
    print(f"Columna '{col}' preservada sin capping superior. Máximo actual: {df[col].max()}")
```

Por el bien de la visualización, para que los gráficos (boxplots, histogramas) sean legibles sin que los extremos aplasten la visualización, vamos a usar la escala logarítmica en los ejes, pero sin modificar los datos subyacentes.

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for i, col in enumerate(['views', 'likes', 'dislikes', 'comment_count']):
    # Usamos el logaritmo para la visualización, pero los datos originales quedan intactos
    sns.boxplot(data=df, y=np.log1p(df[col]), ax=axes[i])
    axes[i].set_title(f'Log({col})')
plt.tight_layout()
plt.show()
```
