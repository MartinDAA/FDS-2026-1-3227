```python

# Encontrar filas que tienen valores nulos

filas_son_nulas = df[df.isnull().any(axis=1)]  # Con esto accedemos a las filas que tengan algún valor nulo en algunas de sus columnas

filas_son_nulas
```

```python
# Encontrar la cantidad total de filas con valores nulos

print(df.isnull().sum().sum()) # Total de valores nulos
```

```python
#Encontrar la cantidad total de filas con valores nulos

print(df.isnull().sum()) # Total de valores nulos por columna


# Con esto podemos darnos cuenta de que la mayoría de valores nulos se encuentra en la columna "description"
```

```python
#La única columna cuyos valores no se deberian repetir son las de la columna "video_id"



conteo = df['video_id'].value_counts()
print(conteo.to_dict())
```

```python


lista= []

for valor,cantidad in conteo.items():


    if cantidad > 1:

        lista.append(df.index[df['video_id'] == valor])



# Se consigue una lista con todas las posiciones de los valores repetidos



```

```python

for i in range(len(lista)):


    for j in range(1,len(lista[i])):

        df.drop(lista[i][j],inplace=True)
        




    
```

```python
conteo2 = df['video_id'].value_counts()
```

```python
conteo2 #Se puede observar que ya no hay valores repetidos
```

```python
# Tomando en cuenta que un video puede NO tener una descripción, cambiaremos sus valores nulos a un valor "sin descripcion"

df['description'] = df['description'].fillna("Sin descripcion")

print(df.isnull().sum())
```

```python
#Como la columna "thumbnail_link" no aporta mucho al análisis que estamos haciendo, optamos por borrarlo


df.drop(columns='thumbnail_link',inplace=True)
```
